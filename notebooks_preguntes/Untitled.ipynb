```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("PokemonData.csv")
```

```python
df['FormaAlterna'] = df.duplicated('Num', keep='first')

b = df[~df['FormaAlterna']]
b.loc[b.groupby('Generation')['Total'].idxmax(), ['Generation','Name','Total']]
```
